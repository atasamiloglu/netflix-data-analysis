```python
import pandas as pd
netflix_dataset = pd.read_csv("netflix_titles.csv")
netflix_dataset.head()
```

```python
netflix_dataset.info()
```

```python
## 1. Content Type Distribution
```

```python
netflix_dataset.value_counts("type",normalize=True) * 100
```

```python
## 2. Top 10 Countries by Content Count
```

```python
netflix_dataset.value_counts("country").head(10)
```

```python
## 3. Analysis of Productions from Turkey
```

```python
turkey_df = netflix_dataset[netflix_dataset['country'] == 'Turkey']
turkey_df.value_counts("type")
```

```python
## 4. Most Recent Turkish Productions
```

```python
turkey_df[["type","title","release_year"]].sort_values("release_year",ascending=False).head(5)
```

```python
## 5. Temporal Analysis: Content Growth Over the Years
```

```python
netflix_dataset["date_added"] = pd.to_datetime(netflix_dataset["date_added"].str.strip())
netflix_dataset["year_added"] = netflix_dataset["date_added"].dt.year
netflix_dataset["year_added"].value_counts().head(3)
```

```python
## 6. Popular Genres and Categories Analysis
```

```python
netflix_dataset['listed_in'].dropna().str.split(', ').explode().value_counts().head(5)
```
